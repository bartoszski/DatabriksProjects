```python
# from the repo folder or add it to PYTHONPATH
import carelink_client2
import pandas as pd
from datetime import datetime
```

```python


client = carelink_client2.CareLinkClient(tokenFile="logindata.json")
if not client.init():
    raise RuntimeError("CareLink init failed")

# recentData is a dict with pump+sensor status and ~24h history
recent = client.getRecentData()

# Example: flatten key pieces (adjust fields to what you need)
events = []

# Sensor readings (5-min buckets)
for e in recent.get("sgs", []):
    events.append({
        "ts": datetime.fromisoformat(e["datetime"].replace("Z","+00:00")),
        "type": "sg",
        "sg_mgdl": e.get("sg"),
        "isig": e.get("isig")
    })

# Boluses (manual + auto corrections)
for b in recent.get("boluses", []):
    events.append({
        "ts": datetime.fromisoformat(b["datetime"].replace("Z","+00:00")),
        "type": "bolus",
        "insulin_u": b.get("insulin"),
        "subtype": b.get("bolusType")  # e.g., AUTO_CORRECTION
    })

# Basal (programmed / auto basal)
for br in recent.get("basal", []):
    events.append({
        "ts": datetime.fromisoformat(br["datetime"].replace("Z","+00:00")),
        "type": "basal",
        "rate_u_per_h": br.get("rate"),
        "auto": br.get("isAutoMode")
    })

df = pd.DataFrame(events).sort_values("ts").reset_index(drop=True)
print(df.tail())
```
